# Simple Calculator

```python
ValueOne = int(input("Enter a value: "))
ValueTwo = int(input("Enter a value: "))

Addition = ValueOne + ValueTwo
subtarction = ValueOne - ValueTwo
multiplication = ValueOne * ValueTwo
division = ValueOne / ValueTwo

print("The addition of two values is: ",Addition)
print("The subtarction of two values is: ",subtarction)
print("The multiplication of two values is: ",multiplication)
print("The division of two values is: ",division)
```
