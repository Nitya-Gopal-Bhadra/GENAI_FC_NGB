## Logical Ops

```python
a = 23
b = 26
c = 3

print (a > b)  # This will print False since 23 is not greater than 26
print (a < b)  # This will print True since 23 is less than 26

c = a<b and a>c
print (c)  # This will print True since both conditions are True

d = a<b or a>c
print (d)  # This will print True since at least one condition is True  

```
