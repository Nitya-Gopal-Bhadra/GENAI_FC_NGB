## List Indexing

```python
a_list = [1,2,3,4,5]
#this is how you index a list in python from 0 to infinite
print (a_list[0])

#now lets see how to see length of a list
print (len(a_list))
```

```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango","orange", "pear", "peach", "grape", "melon", "watermelon"]

# task print only cherry 🍒
print (fruits[2])
```

#### Negetive Indexing 

```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango","orange", "pear", "peach", "grape", "melon", "watermelon"]
print (fruits[-1])
# for complete lists we can use ":"
print(fruits[:])
```

#### List Slicing

```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango","orange", "pear", "peach", "grape", "melon", "watermelon"]
#this is how you index a list in python from 0 to infinite
# Lets slice from cherry to mango
# rember first domain is fine but range should = range + 1
print (fruits[2:5])
```

```python
# Reverse the list
print(fruits[::-1])
```

## List Mutability
### we can replace the value present at specific index in list

```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango","orange", "pear", "peach", "grape", "melon", "watermelon"]

fruits[3] = "Avocado"

print (fruits[:])
```

## Iterating lists by loop

```python
for cart in fruits :
    print(cart)
```
